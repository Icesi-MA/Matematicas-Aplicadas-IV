#Sistemas masa resorte

Los sistemas masa-resorte, son sistemas que están presentes en muchas situaciones prácticas; consideraremos una masa $m$ unida a uno de los extremos de un resorte de longitud $l$ mientras que el otro extremo está sujeto a una pared fija, como muestra la siguiente figura.

![](https://drive.google.com/uc?id=1QfJOs37S-OnULEqWsgg9H3cDzyO2tM9h)



De acuerdo a la ley de Hooke, el resorte ejerce una fuerza de restitución, $F_r$, la cual se opone a la dirección del alargamiento o compresión del resorte y es proporcional al desplazamiento, $s$, es decir $F_r= ks$, $k$ es una constante y es la constante del resorte. De acuerdo a la segunda ley de Newton

$$ \sum \text{fuerzas} = ma $$

asi

$$ m\dfrac{d^2x}{dt^2} = W - F_r = mg -k(s+x) = mg -kx -ks. \ \ \ \ (1) $$

En este caso se han omitido fuerzas externas, es decir el **movimiento es libre y no amortiguado**. En la posición de equilibrio $F_r = W$, de donde

$$ mg = ks $$

reemplazando en $(1)$, tenemos

$$ m\dfrac{d^2x}{dt^2} = mg -kx - mg= -kx, $$

y la ecuación del moviemiento libre no amortiguado está dada por

$$ mx^{\prime\prime} = -kx $$

o

$$ x^{\prime\prime} = -\dfrac{k}{m}x. $$

**Nota:**

Como el peso de la masa induce un movimiento hacia abajo, entonces vamos a considerar todo desplazamiento que esté por debajo de la posición de equilibrio con signo positivo y lo que este por encima de la posición de equilibrio como negativo, ver siguiente figura

![](https://drive.google.com/uc?id=1APfWaSi0OJ1jZo9fNJYZdt2gN9MC30Mz)



**Movimiento libre amortiguado**

En este caso asumimos que existe una fuerza adicional que retarda o amortigua el movimiento de la masa $m$, esto podría ser un recipiente con algún líquido, ver siguiente figura

![](https://drive.google.com/uc?id=1eTm9nPmyVrqbrh11OwNzJBMgpMX2zlcu)

Las fuerzas de amortiguamiento que actúan sobre un cuerpo son proporcionales a la velocidad instántanea y tienen la misma dirección que la fuerza del resorte, entonces en la ecuación diferencial planteada para el movimiento libre no amortiguado tendriamos que considerar el término $\beta\dfrac{dx}{dt}$ de la siguiente forma

$$ m\dfrac{d^2x}{dt^2} = W - F_r -F_a= mg -k(s+x) - \beta\dfrac{dx}{dt}= mg -kx -ks - \beta\dfrac{dx}{dt}$$

simplificando llegamos a

$$ x^{\prime\prime} = -\dfrac{k}{m}x - \dfrac{\beta}{m} x^{\prime} $$



**Moviento forzado amortiguado**


Para este caso se considera una fuerza externa, $f(t)$, que actua sobre la masa, esta fuerza podría ser una fuerza de impulso que iniciaría el movimiento, ver figura


![](https://drive.google.com/uc?id=1ImYOrVziEBWycxeQtegbf49ZDQWgKcvt)

en este caso la ecuación toma la forma

$$ mx^{\prime\prime} = -kx - \beta x^{\prime} + f(t) $$
