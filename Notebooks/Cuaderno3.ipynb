## Campos de direcciones

Encontrar una fórmula cerrada o una expresión para la solución $y=y(t)$ de la ecuación diferencial

$$ \dfrac{dy}{dt}=f(t,y) $$

es bastante complicado y en muchas ocaciones no es posible obtener dicha fórmula. En estas ocasiones podemos usar el concepto geométrico (tangencia) de la derivada de una función para obtener información del comportamiento de las soluciones de la ecuación diferencial. Dado un punto $(t_0,y_0)$ del plano $ty$ en el dominio de la función $f$ tenemos que la pendiente de la recta tangente a la gráfica de la solución $y=y(t)$ de la ecuación diferencial tiene pendiente $m=\frac{dy}{dt}(t_0,y_0)$, usaremos esta interpretación geométrica para construir lo que se conoce como campo de direcciones. Dada una ecuación diferencial, un campo de direcciones es un fragmento del plano $ty$ en el cual dado un punto $(t_0,y_0)$ en este plano se ha graficado un segmento de línea recta el cual corresponde a la recta tangente a la gráfica de la solución $y=y(t)$ que pasa por este punto $(t_0,y_0)$. Para la ecuación diferencial

$$ \dfrac{dy}{dt} = 0 $$

tenemos que para cualquier punto $(t,y)$ la recta tangente tiene pendiente $0$, esto debido a que $f(t,y)=0$ para todo punto $(t,y)$, es decir que las rectas tangentes son rectas paralelas al eje $t$, dibujando segmentos de rectas de igual longitud en cada punto $(t,y)$ tendriamos que el campo de direcciones es
