## El sistema depredador-presa

Comenzamos nuestro estudio de los sistemas de ecuaciones diferenciales considerando dos versiones del modelo depredador-presa. Aquí $R(t)$ denota
la población (en miles, millones o lo que sea) de presas presentes en el tiempo $t$ y  $F(t)$ denota la población de depredadores. Suponemos que tanto $R(t)$ como $F(t)$ son no negativos. Un sistema de ecuaciones diferenciales que podría gobernar los cambios en la población de estas dos especies es:

\begin{align*}
\frac{dR}{dt} &=  2R - 1.2RF \\
\frac{dF}{dt} &=  -F + 0.9RF
\end{align*}

Interprete la presencia de los términos del lado derecho del sistema de ecuaciones.

La presencia de los términos $RF$ en estas ecuaciones hace que este sistema sea difícil de resolver, ¿Qué tipo de ecuación diría usted define este termino? Es imposible derivar fórmulas explícitas para la solución general, pero hay algunas condiciones iniciales que dan soluciones simples.

Intente determinar si existe una solución de equilibrio de este sistema. Por analogía con las ecuaciones de primer orden, llamamos
todo par de funciones constantes una solución de equilibrio para el sistema. El equilibrio $R=F=0$ tiene mucho sentido: si tanto la población de depredadores como la de presas desaparecen, ciertamente no espere que las poblaciones crezcan en un momento posterior.

¿Hay otro posible equilibrio?

## Diagramas de Fase

Como en el caso de  una ecuación diferencial de primer orden, para sistemas de ecuaciones diferenciales existe la posibilidad de estudiar cualitativamente las soluciones mediante los denominados diagramas de fase. Dados $R(t), F(t)$ y un valor de $t$, podemos formar la
pareja $(R(t), F(t))$ y considérarlo como un punto en el plano $RF$. En otras palabras, las coordenadas del punto son los valores de las dos poblaciones en el tiempo $t$. Como $t$ varía, el par $(R(t), F(t))$ traza una curva en el plano $RF$. Esta curva es la curva solución determinada por la condición inicial. Las coordenadas de cada punto de la curva, son las poblaciones de presas y depredadores en el tiempo asociado $t$, y el punto $(R_0, F_0)$
que corresponde a la condición inicial para la solución a menudo se denomina condición inicial de esta curva solución.

El plano $RF$ se denomina plano de fase. Podemos trazar muchas curvas de soluciones en el plano de fase simultáneamente, con lo cual se dibuja el retrato completo de la fase de nuestro sistema depredador-presa

Las soluciones de equilibrio son soluciones que son constantes y, en consecuencia, producen curvas solución $(R(t), F(t))$. En otras palabras,
las curvas  solución que corresponden a soluciones de equilibrio son realmente solo puntos, y nos referimos a ellos como puntos de equilibrio

Ahora consideramos una modificación del modelo depredador-presa en el que asumimos que en la ausencia de depredadores, la población de presas obedece a un modelo logístico más que uno exponencial. Uno de esos modelos para esta situación es el sistema:

\begin{align*}
\frac{dR}{dt} &=  2R\left(1-\frac{R}{2}\right) - 1.2RF \\
\frac{dF}{dt} &=  -F + 0.9RF
\end{align*}

## Geometría de sistemas: Campo vectorial del sistema depredador-presa

Usando vectores, construimos una representación geométrica de un sistema de ecuaciones diferenciales, similar al campo de pendientes, dicha
representación de una ecuación diferencial nos da una manera conveniente de entender sus soluciones.

En el sistema depredador-presa

\begin{align*}
\frac{dR}{dt} &=  2R - 1.2RF \\
\frac{dF}{dt} &=  -F + 0.9RF
\end{align*}

consideramos el   par $(R(t ), F(t))$ como una función vectorialen el plano  $RF$. Para cada $t$, sea $P(t)$ el  vector:

$$P(t)=\begin{bmatrix}
R(t)\\
F(t)\end{bmatrix}$$

de esta manera el sistema presa-predador se escribe

$$\frac{dP(t)}{dt}=\begin{bmatrix}
\frac{dR}{dt}\\
\frac{dF}{dt}\end{bmatrix}=\begin{bmatrix}
2R- 1.2RF\\
  -F + 0.9RF\end{bmatrix}$$

Las ventajas de la notación vectorial comienzan a ser evidentes una vez que consideramos el lado derecho de este sistema como un campo vectorial. El lado derecho del sistema presa-predador es una función que asigna un vector a cada punto en el plano $RF$. Si nosotros denotamos esta función usando el vector $V$, tenemos


$$V\begin{pmatrix}
R\\
F\end{pmatrix}=\begin{bmatrix}
2R- 1.2RF\\
  -F + 0.9RF\end{bmatrix}$$

Podemos dibujar el campo vectorial $V$ colocando el vector $V(P)$ en el punto $P$ correspondiente en el plano $RF$. Calcular $V(P)$ para muchos valores diferentes de $P$ y dibujarlos es un trabajo tedioso para un ser humano, pero es justo el tipo de trabajo que las computadoras  nos ayudan a hacer.  En general, visualizamos este campo vectorial como un "campo" de flechas,
uno basado en cada punto en el plano de $RF$.

## Sistemas de ecuaciones diferenciales de primer orden

Un sistema de ecuaciones diferenciales de primer orden donde la variable independiente es $t$ y las variables dependientes son $x_1$, $x_2$, ..., $x_n$, es un conjunto de ecuaciones diferenciales que tiene la forma

$$f_1(t, x_1, x_2, ..., x_n, x_1^{\prime},x_2^{\prime}, ..., x_n^{\prime})=0$$
$$f_2(t, x_1, x_2, ..., x_n, x_1^{\prime},x_2^{\prime}, ..., x_n^{\prime})=0$$
$$.$$
$$.$$
$$.$$
$$f_n(t, x_1, x_2, ..., x_n, x_1^{\prime},x_2^{\prime}, ..., x_n^{\prime})=0$$

**Nota:** Si solamente se estan considerando 2 variables dependientes, se suelen usar las variables $x$ y $y$, es decir no es necesario hacer uso de subindices. Igual si tenemos 3 variables dependientes, en este caso se suele usar las variables, $x$, $y$ y $z$.

## Forma matricial de un sistema de ecuaciones diferenciales de primer orden

Si definimos el vector columna $X=X(t)$ como

$$X=\begin{bmatrix}
x_1(t) \\
x_2(t) \\
\vdots\\
x_n(t)
\end{bmatrix}$$

entonces

$$\dfrac{dX}{dt}=X^{\prime}=\begin{bmatrix}
x_1^{\prime}(t) \\
x_2^{\prime}(t) \\
\vdots\\
x_n^{\prime}(t)
\end{bmatrix}$$

es decir, la derivada de una función vectorial es la derivada de cada componente. Ahora si, A es la matriz de coeficientes

$$A=\begin{bmatrix}
a_{11}(t) & a_{12}(t) &  \cdots & a_{1n}(t) \\
a_{21}(t) & a_{22}(t) &  \cdots & a_{2n}(t) \\
\vdots & \vdots & \dots & \vdots\\
a_{n1}(t) & a_{n2}(t) &  \cdots & a_{nn}(t)
\end{bmatrix}$$

y $b$ el vector de términos independientes

$$b=\begin{bmatrix}
b_1(t) \\
b_2(t) \\
\vdots\\
b_n(t)
\end{bmatrix},$$

entonces el sistema lineal

$$ \dfrac{dx_1}{dt} = a_{11}(t)x_1 + a_{12}(t)x_2 + \cdots + a_{1n}(t)x_n + b_1(t) $$

$$ \dfrac{dx_2}{dt} = a_{21}(t)x_1 + a_{22}(t)x_1 + \cdots + a_{2n}(t)x_n + b_2(t) $$
$$.$$
$$. $$
$$.$$
$$ \dfrac{dx_n}{dt} = a_{n1}(t)x_1 + a_{n2}(t)x_1 + \cdots + a_{nn}(t)x_n + b_n(t) $$

toma la forma matricial

$$ X^{\prime} = A(t)X + b(t) $$

Si el vector $b(t)$, es tal que $b(t)=0$ para todo $t$, entonces diremos que el sistema de ecuaciones diferenciales lineal es homogéneo, $b(t)\neq 0$ para algún valor de $t$, entonces diremos que el sistema en no homogéneo.
