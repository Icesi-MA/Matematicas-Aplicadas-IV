#Estabilidad del origen para sistemas canónicos.

En este cuaderno estudiaremos la estabilidad del punto crítico $(0,0)^T$ para sistemas canónicos.

El primer sistema a considerar es

$$ X^{\prime} = J X $$

donde $J$ es una matriz $2 \times 2$ dada por

$$ J=\begin{bmatrix}\lambda_1 & 0 \\ 0 & \lambda_2\end{bmatrix} $$

esta es una matriz diagonal y sus valores propios están dados por $\lambda_1$  y $\lambda_2$. Para el valor propio $\lambda_1$ tenemos que la matriz aumentada del sistema para encontrar vectores propios es

$$ \begin{bmatrix} 0 & 0 & | & 0 \\ 0 & \lambda_2-\lambda_1 & | & 0\end{bmatrix} $$

si $\lambda_1 \neq \lambda_2$, tenemos dos valores propios reales diferentes y estamos en el caso $I$ y de la última fila tenemos que $v_2=0$, por tanto un vector propio asociado a este valor propio es

$$ V = \begin{bmatrix}1 \\ 0 \end{bmatrix} $$

y la solución correspondiente es

$$ X_1 = Ve^{\lambda_1t} = \begin{bmatrix}e^{\lambda_1t} \\ 0 \end{bmatrix}.$$

Para el valor propio $\lambda_2$, tenemos la matriz aumentada

$$ \begin{bmatrix} \lambda_1 - \lambda_2 & 0 & | & 0 \\ 0 & 0 & | & 0\end{bmatrix} $$

de donde $v_1=0$, por tanto un vector propio asociado sería

$$ V = \begin{bmatrix}0 \\ 1 \end{bmatrix} $$

y la solución correspondiente es

$$ X_2 = Ve^{\lambda_2t} = \begin{bmatrix} 0 \\ e^{\lambda_2t} \end{bmatrix}.$$

La solución general en este caso esta dada por

$$ X =  \begin{bmatrix} x \\ y \end{bmatrix}c_1X_1 + c_2X_2 = c_1\begin{bmatrix}e^{\lambda_1t} \\ 0 \end{bmatrix}+c_2\begin{bmatrix} 0 \\ e^{\lambda_2t} \end{bmatrix}= \begin{bmatrix} c_1e^{\lambda_1t} \\ c_2e^{\lambda_2t} \end{bmatrix}.$$

Note que las ecuaciones paramétricas, están dada por

$$ x = c_1e^{\lambda_1t} \ \ \ \ \text{y} \ \ \ \ y= c_2e^{\lambda_2t} $$

El único caso en donde se tiene estabilidad asintotica es cuando ambos valores propios son números negativos, esto implicaría que

$$ \lim_{t \to \infty} x(t) = \lim_{t \to \infty} y(t) = 0. $$

Si alguno de los valores propios es positivo entonces no tendríamos los límites anteriores y por lo menos alguna de las soluciones tiene comportamiento infinito. El plano fase queda determinado por la eliminación de parametro, la cual es

$$ y = c_2\bigg(\dfrac{x}{c}\bigg)^{\frac{\lambda_2}{\lambda_1}} $$

Si

$$ J=\begin{bmatrix}\lambda_1 & 0 \\ 0 & \lambda_1\end{bmatrix} $$

esta es una matriz diagonal con un solo valor propio $\lambda_1$. Para este valor propio tenemos que la matriz aumentada del sistema para encontrar vectores propios es

$$ \begin{bmatrix} 0 & 0 & | & 0 \\ 0 & 0 & | & 0\end{bmatrix} $$

Cada columna de la matriz de coeficientes es no pivotal, esto implica que

$$ V = \begin{bmatrix}v_1 \\ v_2 \end{bmatrix} $$

lo cual implica dos vectores propios linealmente independientes $V_1=e_1$ y $V_2=e_2$, por lo tanto

$$ X_1 = \begin{bmatrix}e^{\lambda t} \\ 0 \end{bmatrix}$$

y

$$ X_2 = \begin{bmatrix} 0 \\ e^{\lambda t} \end{bmatrix}.$$

La solución general en este caso esta dada por

$$ X =  \begin{bmatrix} x \\ y \end{bmatrix} = c_1X_1 + c_2X_2 = c_1\begin{bmatrix}e^{\lambda_1 t} \\ 0 \end{bmatrix}+c_2\begin{bmatrix} 0 \\ e^{\lambda_1 t} \end{bmatrix}= \begin{bmatrix} c_1e^{\lambda_1 t} \\ c_2e^{\lambda_1 t} \end{bmatrix}.$$

Las ecuaciones paramétricas, están dadas por

$$ x = c_1e^{\lambda_1 t} \ \ \ \ \text{y} \ \ \ \ y = c_2e^{\lambda_1 t} $$

eliminando paramétro llegamos a

$$ y = \dfrac{c_2}{c_1}x. $$

Como en el caso anterior se tiene estabilidad asintotica solamente si $\lambda_1 < 0$.

Si

$$ J=\begin{bmatrix}\lambda & 1 \\ 0 & \lambda\end{bmatrix} $$

esta es una matriz triangular superior con un solo valor propio $\lambda$. Para este valor propio tenemos que la matriz aumentada del sistema para encontrar vectores propios es

$$ \begin{bmatrix} 0 & 1 & | & 0 \\ 0 & 0 & | & 0\end{bmatrix} $$

La primera fila implica que $v_2=0$ y

$$ V = \begin{bmatrix}v_1 \\ 0 \end{bmatrix} $$

solamente se tiene un solo vector propio, por lo tanto

$$ X_1 = \begin{bmatrix}e^{\lambda t} \\ 0 \end{bmatrix}$$

Ahora, es necesario encontrar el vector propio generalizado $P$, para esto resolvemos el sistema $[A - \lambda I] P = V$, cuya matriz aumentada es

$$\begin{bmatrix} 0 & 1 & | & 1 \\ 0 & 0 & | & 0\end{bmatrix} $$

de donde, $p_2=1$, asi

$$ P=\begin{bmatrix}p_1 \\ 1 \end{bmatrix} $$

si $p_1=0$, entonces un  vector propio generalizado es

$$ P = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$

y

$$ X_2 = Vte^{\lambda t} + Pe^{\lambda t}. $$

La solución general en este caso esta dada por

$$ X =  \begin{bmatrix} x \\ y \end{bmatrix} = c_1X_1 + c_2X_2 = c_1\begin{bmatrix}e^{\lambda t} \\ 0 \end{bmatrix}+c_2\begin{bmatrix} te^{\lambda t} \\ e^{\lambda t} \end{bmatrix}.$$

Las ecuaciones paramétricas, están dadas por

$$ x = c_1e^{\lambda t} +c_2e^{\lambda t} \ \ \ \ \text{y} \ \ \ \ y= c_2e^{\lambda t} $$

igual que en el caso anterior el único caso donde se tendría estabilidad asintótica es si $\lambda < 0$.



Si

$$ J=\begin{bmatrix}a & b \\ -b & a\end{bmatrix} $$

El polinomio caracteristico de esta matriz es

$$ P(\lambda) = (a-\lambda)^2 + b^2 $$

igualando a $0$ y resolviendo para $\lambda$ tenemos los valores propios complejos $\lambda = a \pm bi$, tomaremos $\lambda = a+bi$ para obtener los vectores $B_1$ y $B_2$ del caso III, el sistema de ecuaciones es

$$-biv_1 +bv_2 =0$$
$$-bv_1 -biv_2 = 0$$

multiplicamos la primer ecuación por $i$ y sumandola a la segunda ecuación se obtiene $0=0$ y de la primera ecuación se tiene $v_2 = v_1i$, asi

$$ V = \begin{bmatrix}v_1 \\ v_1i \end{bmatrix} = \begin{bmatrix}v_1 \\ 0 \end{bmatrix} + \begin{bmatrix} 0 \\ v_1 \end{bmatrix}i $$

tomando $v_1 = 1$, tenemos

$$ B_1 = \begin{bmatrix} 1 \\ 0 \end{bmatrix} \ \ \ \ \text{y} \ \ \ \ B_2= \begin{bmatrix} 0 \\ 1 \end{bmatrix} $$

reemplazando en las fórmulas del caso III, tenemos

$$ X_1 = B_1e^{at}\cos{(bt)}-B_2e^{at}\sin{(bt)}=\begin{bmatrix} e^{at}\cos{(bt)} \\ 0 \end{bmatrix} - \begin{bmatrix} 0 \\ e^{at}\sin{(bt)} \end{bmatrix} = \begin{bmatrix} e^{at}\cos{(bt)} \\ - e^{at}\sin{(bt)}\end{bmatrix}$$

y

$$ X_2 = B_2e^{at}\cos{(bt)} + B_1e^{at}\sin{(bt)} = \begin{bmatrix} e^{at}\sin{(bt)} \\ e^{at}\cos{(bt)} \end{bmatrix} $$

La solución general en este caso esta dada por

$$ X =  \begin{bmatrix} x \\ y \end{bmatrix} = c_1\begin{bmatrix} e^{at}\cos{(bt)} \\ - e^{at}\sin{(bt)}\end{bmatrix} + c_2\begin{bmatrix} e^{at}\sin{(bt)} \\ e^{at}\cos{(bt)} \end{bmatrix}=\begin{bmatrix} c_1e^{at}\cos{(bt)}+c_2e^{at}\sin{(bt)} \\ -c_1e^{at}\sin{(bt)}+c_2e^{at}\cos{(bt)} \end{bmatrix}.$$

Este caso es diferente a los anteriores por la presencia del término $e^{\lambda t}$ y los términos trigonométricos. Es claro que si $\lambda < 0$, cada una de las componentes del vector solución tienden a $0$, esto implicaría estabilidad asintótica. Si $\lambda > 0$ cada componente tendería a $\infty$ o $-\infty$ y no se tendría estabilidad. En el caso $\lambda = 0$, el término exponencial no aparecería y en este caso se tendría una estabilidad orbital.
