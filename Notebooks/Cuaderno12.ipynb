## Linealización de un sistema no lineal plano en torno al equilibrio $(a,b)$

De acuerdo a lo estudiado en aplicadas 2, cada función derivable de dos variables se puede linealizar alrededor de un punto donde esta sea derivable y la linealización para una función $f$ en el punto $(a,b)$ es

$$ L_f(x,y) = f(a,b) + f_x(a,b)(x-a) + f_y(a,b) $$

y para $g$

$$ L_g(x,y) = g(a,b) + g_x(a,b)(x-a) + g_y(a,b). $$

Si $(a,b)$ es un punto critico para el sistema

$$ \dfrac{dx}{dt} = f(x,y) $$
$$ \dfrac{dy}{dt} = g(x,y) $$

se tendría que, $f(a,b)=g(a,b)=0$, por lo tanto la linealización toma la forma

$$L_f(x,y) = f_x(a,b)(x-a) + f_y(a,b)(y-b) $$
y
$$L_g(x,y) = g_x(a,b)(x-a) + g_y(a,b)(y-b),$$

Como sabemos, si el punto $(x,y)$ está suficientemente cerca del punto $(a,b)$, entonces

$$f(x,y) \approx f_x(a,b)(x-a) + f_y(a,b)(y-b) $$
y
$$g(x,y) \approx g_x(a,b)(x-a) + g_y(a,b)(y-b),$$

así, que una aproximación lineal al sistema $(1)$ alrededor del punto crítico $(a,b)$ es

$$ \dfrac{dx}{dt} = f_x(a,b)(x-a) + f_y(a,b)(y-b) $$
$$ \dfrac{dy}{dt} = g_x(a,b)(x-a) + g_y(a,b)(y-b) $$

Note que este es un sistema lineal homogéneo con coeficientes constantes, en forma matricial

$$\begin{bmatrix} x \\ y \end{bmatrix}^\prime = \begin{bmatrix}f_x(a,b) & f_y(a,b) \\ g_x(a,b) & g_y(a,b)\end{bmatrix}\begin{bmatrix} x -a \\ y -b \end{bmatrix}$$

el sistema anterior es la linealización del sistema no lineal alrededor del punto crítico $(a,b)$. La matriz del sistema linealizado se conoce como la matriz Jacobiana.

$$J(a,b) = \begin{bmatrix}f_x(a,b) & f_y(a,b) \\ g_x(a,b) & g_y(a,b)\end{bmatrix}$$
