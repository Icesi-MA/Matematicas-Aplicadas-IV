#Ecuaciones diferenciales lineales de primer orden

Una ecuación diferencial lineal de primer orden es una ecuación diferencial que se puede escribir de la forma

\begin{equation} \dfrac{dy}{dt} + p(t)y = q(t) \ \ \ \ \ \ \ \ (1)\end{equation}

**Algunos ejemplos de ecuaciones diferenciales lineales de primer orden y otras que no lo son**

**Ejemplo 1**

$$ \dfrac{dy}{dt} + y = 0 $$

Note que en este caso $p(t)=1$ y $q(t)=0$, son funciones constantes y la ecuación diferencial es lineal y también separable.


**Ejemplo 2**

$$ t\dfrac{dy}{dt} + y = 1 $$

Es una ecuación diferencial lineal con $p(t)=q(t)=1/t$.

**Ejemplo 3**

$$ \dfrac{dy}{dt} + ty^2 = t^2 $$

No es una ecuación diferencial lineal, ya que si tratamos de escribirla de la forma (1), se tendría

$$ \dfrac{dy}{dt} + tyy = t^2 $$

lo cual implicaría que la función $p=p(t,y)=ty$ y $q(t)=t^2$ lo cual no es posible ya que las funciones $p$ y $q$ deben depender solamente de la variable independiente, en este caso $t$.



#El caso homogéneo.

Si $q(t)=0$ para todo $t$, diremos que la ecuación diferencial (1) es homogénea, es decir una ecuación diferencial lineal homogenea es una ecuación diferencial que se puede escribir de la forma

$$ \dfrac{dy}{dt} + p(t)y = 0 $$

Note que esta es una ecuación diferencial separable y su solución está dada por

$$ y = y(t) = Ce^{-\int p(t)dt} $$

**Ejercicio 2**

La ecuación diferencial

$$\dfrac{dy}{dt}=0$$

es lineal homogénea, aplique la última fórmula para encontrar su solución general.

#El caso no homogéneo

Si $q(t) \neq 0$, diremos que la ecuación diferencial lineal de primer orden (1) es no homogénea. En este caso la ecuación diferencial lineal homogénea asociada a esta ecuación es

$$ \dfrac{dy}{dt} + p(t)y = 0. $$

La idea es encontrar la solución general de la ecuación (1), para esto  encontraremos una función $F=F(t)$ (factor integrante), que convierta el lado izquierdo de la ecuación diferencial en la derivada de un producto. Una vez encontrada $F$, multiplicaremos la ecuación diferencial por esta función, es decir

$$F(t)\dfrac{dy}{dt} + p(t)F(t)y = F(t)q(t),$$

tal que el lado izquierdo de la ecuación se convierta en la derivada del producto $F(t)y$, con esto, llegamos a

$$ \dfrac{d}{dt}(F(t)y) = F(t)q(t)$$

integramos con respecto a $t$ ambos lados y llegamos a

$$ F(t)y = \int F(t)q(t)\ dt +C $$

finalmente

$$ y = y(t) = \dfrac{\displaystyle\int F(t)q(t)\ dt}{F(t)} +\dfrac{C}{F(t)} $$

o

$$ y= y(t) = F^{-1}(t)\int F(t)q(t)\ dt + CF^{-1}(t) $$


¿Cómo encontrar la función $F(t)$?

Igualamos

$$ \dfrac{d}{dt}(F(t)y)= F(t)\dfrac{dy}{dt} + p(t)F(t)y $$

aplicamos la regla del producto

$$ F^{\prime}y +Fy^{\prime} = F(t)\dfrac{dy}{dt} + p(t)F(t)y $$

cancelando, llegamos a

$$ F^{\prime}y = p(t)F(t)y $$

por lo tanto

$$ \dfrac{dF}{dt} = p(t)F $$

la cual es una ecuación diferencial separable, separamos

$$ \dfrac{dF}{F} = p(t)$$

integramos

$$ \int\dfrac{1}{F}\ dF = \int p(t)\ dt $$

asi,

$$ \ln{F} = \int p(t)\ dt $$

despejando tenemos

$$ F(t) = e^{\int p(t)\ dt}.$$

**Nota:**

No es necesario considerar en este punto constantes de integración.


Ahora, con el factor integrante tenemos completamente determinada la solución general de la ecuación diferencial lineal

$$\dfrac{dy}{dt} + p(t)y = q(t)$$

la cual es

$$ y = y(t) = Ce^{-\int p(t)\ dt} + e^{-\int p(t)\ dt}\int q(t)e^{\int p(t)\ dt} $$

#Algunos ejemplos de ecuaciones diferenciales lineales

**Ejemplo 4**

Resuelva $$ \dfrac{dy}{dt}=t-y $$

**Sol:**

En este caso transponemos términos y  llegamos a la ecuación diferencial

$$ \dfrac{dy}{dt} + y = t, \ \ \ \ \ (1) $$

la cual tiene la forma de una lineal, con $p(t)=1$, lo que nos lleva al factor integrante

$$ \rho (t) = e^{\int{1\,dt}}= e^t. $$

Multiplicamos la ecuación diferencial (1) por el factor integrante encontrado y llegamos a

$$ e^t\dfrac{dy}{dt} + e^ty = te^t, $$

note (como demostramos en clase) que el lado izquierdo es la derivada con respecto a $t$ del producto $\rho(t)y$, es decir

$$ \dfrac{d}{dt}(e^ty) = te^t $$

para eliminar la derivada del lado izquierdo, integramos con respecto a $t$, lo cual no lleva a

$$ e^ty = \int{te^t\,dt} $$

integramos por partes el lado derecho con la escogencia $u=t$ y $dv=e^tdt$, y llegamos a

$$ e^ty = te^t - e^t + C, $$

despejando obtenemos la solución explícita de la ecuación diferencial original, es decir

$$ y(t) = t -1 + Ce^{-t}. $$

**Ejemplo 5**

Resuelva $$ \dfrac{dy}{dt}-2y=e^t $$

**Sol:**

La ecuación dada tiene la forma lineal, con $p(t)=-2$ y un factor integrante es

$$ \rho (t) = e^{\int{-2\,dt}}= e^{-2t}. $$

Multiplicamos la ecuación diferencial por el factor integrante encontrado y llegamos a

$$ e^{-2t}\dfrac{dy}{dt} -2e^{-2t}y = e^{-t}, $$

el lado izquierdo es la derivada con respecto a $t$ del producto $\rho(t)y$, es decir

$$ \dfrac{d}{dt}(e^{-2t}y) = e^{-t} $$

integramos con respecto a $t$, lo cual no lleva a

$$ e^{-2t}y = \int{e^{-t}\,dt} $$

integrando tenemos

$$ e^{-2t}y = -e^{-t} + C, $$

despejando obtenemos la solución explícita de la ecuación diferencial original, es decir

$$ y(t) = -e^t + Ce^{2t}. $$

**Ejemplo 6**

Resuelva $$ \dfrac{dy}{dt}-2ty =t $$

**Sol:**

En este caso $p(t)=-2t$, esto implica que el factor integrante sea $\rho(t)=e^{\int p(t)dt}= e^{\int{-2t}dt}=e^{-t^2}$. Multiplicamos la e dif por el F integrante

$$ e^{-t^2}\dfrac{dy}{dt}-2te^{-t^2}y =te^{-t^2} $$

$$ \dfrac{d}{dt}(e^{-t^2}y) = te^{-t^2} $$

integramos

$$ e^{-t^2}y = \int te^{-t^2}\,dt = -\dfrac{1}{2}e^{-t^2} + C $$

$$ y(t) = Ce^{t^2} -\dfrac{1}{2}. $$


**Ejemplo 7**

Resuelva $$ \dfrac{dy}{dt}-2ty = 1 $$

**Sol:**

En este caso $p(t)=-2t$, esto implica que el factor integrante sea $\rho(t)=e^{\int p(t)dt}= e^{\int{-2t}dt}=e^{-t^2}$. Multiplicamos la ecuación diferencial por el Factor integrante

$$ e^{-t^2}\dfrac{dy}{dt}-2te^{-t^2}y =e^{-t^2}, $$

el lado izquierdo se convierte en

$$ \dfrac{d}{dt}(e^{-t^2}y) = te^{-t^2} $$

integrando

$$ e^{-t^2}y = \int e^{-t^2}\,dt $$

la integral del lado derecho no es una integral elemental o que se pueda resolver con métodos tradicionales de integración, en este caso la solución queda de la forma

$$ y(t) = \dfrac{\displaystyle\int e^{-t^2}\,dt}{e^{-t^2}} + \dfrac{C}{e^{-t^2}}, $$

por lo tanto

$$ y(t) =  e^{t^2}\int e^{-t^2}\,dt + Ce^{t^2}.$$
