## Ideas fundamentales de ecuaciones diferenciales ordinarias

Las ecuaciones diferenciales ordinarias (EDO) surgen en todo tipo de contextos, pero las más frecuentes y citadas con mayor frecuencia son las de la física y la ingeniería. Las aplicaciones de la Segunda Ley de Newton, por ejemplo, dan ecuaciones diferenciales que relacionan la posición, la velocidad y la aceleración. La segunda ley de Newton es la ecuación
$F=ma$  dónde $F$  es una fuerza que actúa sobre un cuerpo, $m$  es la masa del cuerpo y $a$  es la aceleración. que se puede reformular como
$F=mv'$, una ecuación diferencial con la velocidad $v$ como incógnita, o $F=
ms''$, una ecuación diferencial de segundo orden con la posición $s$ como incógnita.

Algunos de los casos en los que usamos la segunda ley de Newton tienen buenas soluciones analíticas, como el movimiento de un objeto sobre una superficie sin fricción bajo una fuerza constante. Sin embargo, muchos de estos casos están muy idealizados y no reflejan la verdadera realidad. No hace falta mucho para que las ecuaciones diferenciales resultantes de la segunda ley de Newton sean terriblemente difíciles, o quizás imposibles, de resolver. Simplemente agregue algo de fricción, tal vez permita que múltiples cuerpos interactúen, o considere las fuerzas que actúan de manera diferente en múltiples escalas de longitud. Un ejemplo famoso es el problema de los tres cuerpos donde las fuerzas gravitatorias impulsan el movimiento de tres cuerpos celestes. Las ecuaciones diferenciales resultantes no tienen solución analítica y la única manera de mostrar cómo evolucionan las posiciones de los cuerpos en el tiempo es con una aproximación numérica… ¡Y eso es solo 3 cuerpos celestes! ¡Imagine lo complicadas que se vuelven las ecuaciones diferenciales al tratar de modelar nuestro sistema solar (o nuestra galaxia)!

La imposibilidad de resolver una ecuación diferencial se deriva en parte de la imposibilidad de integrar la mayoría de las funciones. Si tuviéramos que elegir al azar funciones para integrar, encontraríamos que la gran mayoría no tiene antiderivadas. La historia en las EDO es la misma: elija cualquier combinación de una función, sus derivadas y otras funciones forzadas y encontrará que no hay forma de llegar a una solución analítica que involucre las operaciones y funciones regulares de las matemáticas: combinaciones lineales, potencias , raíces, funciones trigonométricas, logaritmos, etc. Hay teoremas de ecuaciones diferenciales que garantizan la existencia y unicidad de soluciones para muchas ecuaciones diferenciales, pero saber que la solución existe no es suficiente para ir a buscarla. Las técnicas numéricas nos dan una vía para al menos aproximar estas soluciones.

## Ecuaciones diferenciales de primer orden (lineales vs no lineales)

Una EDO de primer orden conecta una función $y(t)$ con su derivada  $\frac{dy}{dt}$. La tasa de cambio en $y$ es decidida por $y$ (y posiblemente por $t$). Veamos cuatro ejemplos

$$1.\,\frac{dy}{dt}=y,\quad 2.\,\frac{dy}{dt}=-y \quad 3.\,\frac{dy}{dt}=2ty,\quad 4.\,\frac{dy}{dt}=y^2$$

¿Cuáles de las anteriores ecuaciones son lineales en $y$ y $\frac{dy}{dt}$?

A continuación encontraremos la solución a estas cuatro ecuaciones, en estos casos utilizando simplemente integración. Por ejemplo, para el siguiente modelo:

$$\frac{dy}{dt}=f(t).$$

La solución completa es

$$y(t)=\int_0^t f(s)\;ds+C.$$

y una particular viene dada por la condición inicial $y(0)=C$.

Es fácil ver que las soluciones de cada ED vienen dadas por

\begin{align*}
1.\frac{dy}{dt}&=y,\quad y(t)=e^t,\\
2. \frac{dy}{dt}&=-y, \quad y(t)=e^{-t},\\
3.\frac{dy}{dt}&=2ty,\quad y(t)=e^{t^2},\\
4.\frac{dy}{dt}&=y^2,\quad y(t)=\frac{1}{1-t}.
\end{align*}

**Definición (Ecuación diferencial)** Una ecuación diferencial es una ecuación que relaciona la derivada (o derivadas) de una función desconocida consigo misma.

**Definición (Solución de ecuaciones diferenciales)** Una solución de una ecuación diferencial (también llamada solución analítica) es una función que, cuando se sustituye en la ecuación diferencial, crea un enunciado verdadero.

Nuestro especial interés ahora está en las ecuaciones lineales con coeficientes constantes, como 1 y 2. De hecho $$\frac{dy}{dt}=y$$ es la propiedad más importante de la gran función $y=e^t$. El cálculo
tuvo que crear $e^t$ , porque una función del álgebra (como $y = t^n$ ) no puede ser igual a su derivada. Este caso en general corresponde al modelo de crecimiento (decaimiento) $$\frac{dy}{dt}=ay$$

cuya solución viene dada por $y=Ce^{at}.$

**Crecimiento demográfico ilimitado**

Un modelo elemental de crecimiento de la población se basa en el supuesto:

* La tasa de crecimiento de la población es proporcional al tamaño de la población.

Tenga en cuenta que la tasa de cambio de una población depende solo del tamaño de la población y nada más. En particular, se ignoran las limitaciones de espacio o recursos. Debido a que la suposición es tan simple, esperamos que el modelo también lo sea.

Las cantidades involucradas son
$t =$ tiempo (variable independiente),
$P =$ población (variable dependiente), y
$k =$ constante de proporcionalidad (parámetro) entre la tasa
del crecimiento de la población y el tamaño de la población.
El parámetro $k$ a menudo se denomina "coeficiente de tasa de crecimiento".

Ahora expresemos nuestra suposición usando esta notación. La tasa de crecimiento de la población $P$ es la derivada $\frac{dP}{dt}$. Siendo proporcional a la población se expresa como el producto, $kP$, de la población $P$ y la constante de proporcionalidad $k$. De ahí nuestro
la suposición se expresa como una ecuación diferencial similar a la anterior, esto es,

$$\frac{dP}{dt}=kP$$

la cual es uno de los principales ejemplos de una ecuación diferencial lineal de primer orden.

Como $\frac{dP}{dt}=kP$ para alguna constante $k$, $\frac{dP}{dt} = 0$ si $P = 0$. Así, la función constante $P(t) = 0$ es una solución de la ecuación diferencial. Este tipo especial de solución se llama solución de equilibrio porque es constante para siempre. En cuanto al modelo poblacional, corresponde a una especie que es inexistente. ¿Por qué?

Si $P(t_0)  = 0$ en algún momento $t_0$, entonces  $\frac{dP}{dt}=kP(t_0)\not= 0$ en $t_0$. Como consecuencia, la población no es constante. Si $k > 0$ y $P(t_0) > 0$, tenemos  $\frac{dP}{dt}=kP(t_0)>0$,
en el tiempo $t = t_0$ y la población está aumentando (como era de esperar). A medida que $t$ aumenta, $P(t)$ se vuelve más grande, por lo que  $\frac{dP}{dt}$ se vuelve más grande. A su vez, $P(t)$ aumenta aún más rápido.
Es decir, la tasa de crecimiento aumenta a medida que aumenta la población. Por lo tanto, esperamos
que la gráfica de la función $P(t)$ podría parecerse a la Figura anterior, a cuál de las dos curvas?

El valor  $P(t)$ en $t = 0$ es llamado condición inicial. (Leer la sección US Population del texto guía)

**Los enfoques analítico, cualitativo y numérico**

El **enfoque analítico** busca fórmulas explícitas que describan el comportamiento de las soluciones. Desafortunadamente, una gran cantidad de ecuaciones importantes no se pueden manejar con el enfoque analítico; simplemente no hay forma de encontrar una fórmula exacta que describa
la situación. Por lo tanto, nos vemos obligados a recurrir a métodos alternativos.

Un método particularmente poderoso para describir el comportamiento de las soluciones es el **enfoque cualitativo**. Este método implica el uso de la geometría para dar una visión general de la comportamiento del modelo. Este método no da valores precisos de la solución en momentos específicos, pero
a menudo permite determinar el comportamiento a largo plazo de las soluciones.
Con frecuencia, este es justo el tipo de información que necesitamos.

El tercer enfoque para resolver ecuaciones diferenciales es **numérico**. El computador aproxima la solución que buscamos. Las técnicas de aproximación numérica son un poderosa herramienta para darnos intuición sobre las soluciones que deseamos.

**Separación de Variables**

Desafortunadamente, es raro el caso de que podamos encontrar soluciones explícitas de una ecuación diferencial. Muchas ecuaciones diferenciales tienen soluciones que no se pueden expresar en
términos de funciones conocidas como polinomios, exponenciales o funciones trigonométricas. Sin embargo, hay algunos tipos especiales de ecuaciones diferenciales para las que podemos derivar soluciones explícitas, y en esta sesión discutimos uno de estos tipos de diferenciales
ecuaciones. La típica ecuación diferencial de primer orden se da en la forma dada a continuación, podemos recordar una de las técnicas más sencillas para resolver ODE a mano.

**Teorema (separación de variables)** Para resolver una ecuación diferencial de la forma

$$\frac{dx}{dt}=f(x)g(t)$$

podemos separar las variables y reescribir el problema como

$$\int \frac{dx}{f(x)} =\int g(t)\;dt.$$

Integrando ambos lados y resolviendo para $x(t)$ da la solución.

**Ejemplo.** Resuelva la ED de primer orden $\frac{dy}{dt}=2ty$

**Ejemplo (Soluciones faltantes)**

Si es posible separar variables en una ecuación diferencial, parecería que resolver la ecuación se reduce a una cuestión de calcular varias integrales. Esto es cierto en general, pero hay hay algunas trampas ocultas, como muestra el siguiente ejemplo. Considere la ecuación diferencial *autónoma*

$$\frac{dy}{dt}= y^2$$
Esta es una ecuación separable, y su solución es fácil de encontrar. Encúentrela y muestre que hay una solución constante que no pertenece a esa familia de soluciones. Esa solución faltante se denomina un *equilibrio* de la ecuación diferencial, y estudiaremos mas en detalle ese tipo de soluciones mas adelante.
