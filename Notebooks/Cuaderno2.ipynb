```python
#Sumas de Riemann por izquierda
def SumaIzq(f, a, b, n):
    h = float(b-a)/n
    result = 0
    for i in range(n):
        result += f(a + i*h)
    result *= h
    return result
```

## Forma estandar o normal de una ecuación diferencial de primer orden

Una ecuación diferencial de primer orden esta escrita en forma normal o estandar si se puede escribir de la forma

$$ \dfrac{dy}{dt} = f(t,y) $$

**Caso más simple para ecuaciones diferenciales de primer orden**

Si la función de dos variables $f(t,y)$ solamente depende de la variable independiente $t$, tendríamos la ecuación diferencial

$$ \dfrac{dy}{dt} = g(t), $$

en este caso la solución general de la ecuación diferencial se obtiene integrando ambos lados con respecto a la variable independiente

$$ \int\dfrac{dy}{dt}\,dt = \int g(t)\, dt, $$

usando el teorema fundamental del cálculo en el lado izquierdo, tenemos

$$ y = y(t) =  \int g(t)\,dt. $$

**Ejemplo 4**

La solución general de la ecuación diferencial $dy/dt = 8$ está dada por

$$ y(t)=\int 8\ dt =8t +C $$

## **Ecuaciones diferenciales separables**

Diremos que una ecuación diferencial de primer orden

$$ \dfrac{dy}{dt} = f(t,y) $$

es separable, si la función $f$ se puede escribir como el producto de dos funciones $g(t)$ y $h(y)$, es decir

$$ \dfrac{dy}{dt} = g(t)h(y). $$

¿Cómo encontrar su solución general?

Separamos variables

$$ \dfrac{1}{h(y)}\dfrac{dy}{dt} = g(t), \ \ \ \text{siempre que} \ \ \  h(y)\neq 0. $$

La variable independiente aquí es $t$, entonces integramos con respecto a $t$ en ambos lados de la ecuación

$$ \int\dfrac{1}{h(y)}\dfrac{dy}{dt}\ dt = \int g(t)\ dt,$$
note que en la integral del lado izquierdo aparecen dos variables ($y$ y $t$), lo que sigue es tratar de que esta integral quede solamente en términos de una variable, para esto realizamos la sustitución $u=y(t)$, derivando tenemos que $du = \frac{dy}{dt}dt$, con esta sustitución el lado izquierdo toma la forma

$$ \int\dfrac{1}{h(u)}du = \int g(t)dt $$

como $u=y$, entonces

$$ \int \dfrac{1}{h(y)}dy = \int g(t)dt $$

y de esta forma la integral del lado izquierdo esta en términos de $y$ y la del lado derecho en términos de la variable $t$.

Si $H(y)$ es una primitiva o antiderivada de la función $1/h(y)$, en la integral del lado izquierdo, entonces tenemos lo que se conoce como solución implícita de la ecuación diferencial

$$ H(y) = G(t) + C, $$

note que en este punto la variable dependiente $y$ no se ha despejado en términos de la variable independiente $t$, si este despeje es posible, es decir si se llega a la expresión

$$ y = y(t) = H^{-1}(G(t) + C) $$

entonces diremos que la solución de la ecuación diferencial está en forma explícita.

**Ejemplo 5**

Resuelva la ecuación diferencial $$\dfrac{dy}{dt}=-ty$$

**Solución**

La ecuación diferencial es separable, el primer paso es dejar en un lado todo lo que tenga que ver con la variable $y$ y en el otro lado del igual lo que tenga que ver con la variable $t$, en este caso

$$ \dfrac{1}{y}\ dy = -t\ dt $$

ahora, integramos ambos lados (esto es solamente poner el simbolo integral en cada lado)

$$ \int\dfrac{1}{y}\ dy = \int-t\ dt $$

note que los diferenciales son los que me indican con respecto a que variable se debe integrar en cada caso, ahora calculamos las integrales

$$ \ln |y| = -\dfrac{1}{2}t^2 +c_1, $$

a este punto se tiene una solución implícita de la ecuación diferencial. Ahora, tratamos de despejar $y$ para obtener una solución explícita

$$|y| = e^{-\frac{1}{2}t^2 +c_1}=e^{-\frac{1}{2}t^2}e^{c_1}=e^{c_1}e^{-\frac{1}{2}t^2}$$

asi,

$$y=y(t)=\pm e^{c_1}e^{-\frac{1}{2}t^2} $$

por lo tanto la solución general de la ecuación diferencial en forma explícita es

$$ y=y(t)=Ce^{-\frac{1}{2}t^2}, $$

donde hemos realizado el cambio $C=\pm e^{c_1}$.

## Problemas de valor inicial de primer orden

Un problema de valor inicial de primer orden es un problema que involucra una ecuación diferencial diferencial de primer orden

$$ \dfrac{dy}{dt} = f(t,y) $$

junto con una condición del tipo

$$ y(t_0) = y_0. $$

**Ejemplo 6**

Resuelva el problema de valor inicial

$$\left\{ \begin{array}{l} \dfrac{dy}{dt} = -ty, \\ \\ y(0)=-1. \end{array} \right.$$

**Solución:**

Un problema de valor inicial se puede puede resolver en dos pasos:

1. Encontrando la solución general de la ecuación diferencial, como vimos en el último ejemplo tenemos que la solución general está dada por

$$ y(t) = Ce^{-\frac{1}{2}t^2}. $$

2. Ahora usaremos la condición inicial para determinar el valor de la constante $C$

$$ y(0) = Ce^{0} = C = -1 $$

de donde se tiene que $C=-1$, con esto tenemos que la solución del problema de valor inicial es

$$ y(t)= -e^{-\frac{1}{2}t^2} $$

**Ejemplo 7**

Encuentre $x(50)$ si

$$ \frac{dx}{dt} = e^{-t^2}, \ \ \ \ x(0)=2. $$

**Solución:**

Para este problema tenemos los siguiente, en lugar de integrar indefinidamente integraremos sobre el intervalo $[0,50]$ de la siguiente forma

$$ \int_0^{50}\dfrac{dx}{dt}\,dt = \int_0^{50} e^{-t^2}\,dt $$

usando el teorema fundamental del cálculo

$$ x(50) - x(0) = \int_0^{50}e^{-t^2}dt $$

$$ x(50) = \int_0^{50}e^{-t^2}dt + x(0) $$

como $x(0)=2$, entonces

$$x(50) = \int_0^{50}e^{-t^2}dt + 2$$

para el cálculo de la integral definida podemos usa cualquiera de los métodos implementados anteriormente

```python
def g(s):
  return np.e**(-s*s)
```

```python
x = midpoint(g, 0, 50, 1000) + 2
x
```
