## Aplicaciones Ecuaciones diferenciales

**Crecimiento y/o decrecimiento exponencial**

El problema de valor inicial
$$\dfrac{dx}{dt} = kx, \ \ \ \ \  x(t_0) = x_0, \ \ \ \  (1)$$

donde $k$ es una constante de proporcionalidad, se emplea como
modelo de distintos fenomenos en los que intervienen crecimiento,
decrecimiento o desintegración de sustancias

**Dinámicas de población**


Uno de los primeros intentos para modelar matematicamente el
crecimiento poblacional humano fue presentado por el economista
ingles Thomas Malthus, en 1798. La idea de este modelo (modelo
de Malthus) es suponer que la tasa de cambio o ritmo con que la
población de un pais crece en un determinado tiempo es proporcional
al total de la población en ese tiempo, si $P(t)$ representa la población
total en el timepo t entonces tendriamos el modelo

$$\dfrac{dP}{dt} = kP, \ \ \ \  (2) $$


**Nota:** Como $P(t)$ denota población, entonces para todo $t$,
$P(t) > 0$, por lo tanto, si $k > 0$ tendremos crecimiento poblacional
y si $k < 0% entonces se tiene que la población es decreciente.

**Ley de Newton de enfriamiento**

Esta ley nos dice que la velocidad con que la temperatura de un
cuerpo cambia es proporcional a la diferencia que hay entre la
temperatura del cuerpo y la temperatura del medio que lo rodea.
Si $T(t)$ representa la temperatura del cuerpo en el instante $t$ y $T_m$
la temperatura del medio, entonces la ley de Newton de enfriamiento
esta dada por

$$\dfrac{dT}{dt} = k(T - T_m). \ \ \  (3)$$

**Movimiento uniformemente acelerado.**


Supongamos que una partícula se mueve sobre una línea recta con aceleración constante, es decir

$$ a(t) = c $$
donde $c$ es una constante. De los cursos de física aprendimos que la aceleración es la derivada de la función velocidad, entonces tenemos la ecuación diferencial de primer orden

$$ \dfrac{dv}{dt} = c, $$

ahora, la velocidad es la derivada de la función posición, si $x=x(t)$ nos dá la posición de la partícula sobre la recta, entonces

$$ \dfrac{d}{dt}\bigg(\dfrac{dx}{dt}\bigg) = c $$

de donde tenemos la ecuación diferencial de segundo orden

$$ \dfrac{d^2x}{dt^2} = c, $$

equivalentemente

$$ x^{\prime\prime} = c. $$

Este tipo de ecuaciones se pueden resolver apelando al Teorema fundamental del cálculo, es decir con integraciones sucesivas, integramos la primer vez y se tiene que

$$ x^{\prime}(t) = \int c\,dt $$

$$ x^{\prime}(t) = ct + c_1 $$

si se tiene una condición inicial del tipo $v(0)=v_0$, se tiene que $c_1=v_0$, por tanto

$$ x^{\prime}(t) = ct + v_0 $$

integrando de nuevo

$$ x(t) = \dfrac{c}{2}t^2 + v_0t + c_2 $$

con una condición inicial del tipo $x(0)=x_0$, se llega a que $c_2=x_0$ y tendriamos la función posición como sigue

$$ x(t) =  \dfrac{c}{2}t^2 + v_0t + x_0.$$

## Método de Euler

El método de Euler es un método numérico que nos permite aproximar la solución del problema de valor inicial

$$ \dfrac{dy}{dx} = f(x,y), \ \ \ \ \ \ y(x_0)=y_0. $$

Este método de la definición de la derivada, en el curso de matemáticas aplicadas I aprendimos que

$$ f^{\prime}(x_0) = \lim_{h \to 0} \dfrac{\Delta y}{\Delta x} $$

donde $\Delta y = y_1 - y_0$, $y_1 = f(x_1)$ y $\Delta x = x_1 - x_0 = h$. Ahora para un $h$ suficientemente pequeño tenemos

$$ f^{\prime}(x_0) \approx \dfrac{\Delta y}{\Delta x}, $$

de donde

$$ \Delta y \approx f^{\prime}(x_0) \Delta x $$

reemplazando, tenemos

$$ y_1 - y_0 \approx f^{\prime}(x_0)h $$

asi

$$ y_1 \approx y_0 +f^{\prime}(x_0)h $$

Podemos continuar este proceso para determinar $y_2$ y de esta forma tenemos el conocido método de Euler. La iteración del método de Euler es la siguiente

$$ y_{n+1} = y_n + hf(x_n,y_n) $$
$$ x_{n+1} = x_n + h $$

Aquí $h$ es un valor conocido como el tamaño de paso y $n = 0, 1, 2, 3, 4, \dots$
