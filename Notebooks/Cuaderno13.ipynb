## Modelo depredador presa Lotka - Volterra.


Si $x=x(t)$ es la población de la presa en el instante $t$ y $y=y(t)$ es la población del depredador en el instante $t$, entonces el modelo depredador presa de Lotka - Volterra es


$$ \dfrac{dx}{dt} = ax -bxy$$

$$ \dfrac{dy}{dt} = cxy -dy$$

$a, b, c$ y $d$ son  constantes positivas.

**Ejemplo 1.**

$$ \dfrac{dx}{dt} = 0.1x -0.00002xy$$

$$ \dfrac{dy}{dt} = 0.0001xy - 0.04y$$


el sistema que nos dá los puntos críticos es

$$ x(0.1 -0.00002y) = 0 $$

$$ y(0.0001x - 0.04) = 0$$

De (1) se tiene que: $x=0$ o $0.1= 0.00002y$, de donde $y=5000$

De (2) se tiene que: $y=0$ o $0.04 = 0.0001x$, de donde $x = 400$

Los puntos críticos son:

$$ (0,0), (400,5000) $$

Clasificación ptos críticos

En este caso

$$ f(x,y) = 0.1x -0.00002xy $$

y

$$ g(x,y) = 0.0001xy - 0.04y $$

$$ f_x(x,y) = 0.1 -0.00002y $$

$$ f_y(x,y) = -0.00002x $$

$$g_x(x,y) = 0.0001y$$

$$ g_y(x,y)= 0.0001x -0.04 $$

La matriz Jacobiana es

$$ J(x,y) = \begin{bmatrix} 0.1 -0.00002y & -0.00002x \\ 0.0001y & 0.0001x -0.04\end{bmatrix} $$

Entonces la linealización alrededor del pto crítico $(0,0)$ es

$$ \begin{bmatrix} x \\y\end{bmatrix}^{\prime} = \begin{bmatrix} 0.1 & 0 \\ 0 & -0.04\end{bmatrix}\begin{bmatrix}x-0 \\y-0\end{bmatrix} $$

los valores propios son $\lambda =0.1 >0$ y $\lambda = -0.04 <0$, este es un punto silla, el cual es inestable.

la linealización alrededor del pto crítico $(400,5000)$ es

$$ \begin{bmatrix} x \\y\end{bmatrix}^{\prime} = \begin{bmatrix} 0 & -0.008 \\ 0.5 & 0\end{bmatrix}\begin{bmatrix}x - 400 \\y - 5000\end{bmatrix} $$

El polinomio característico es

$$ p(\lambda) = \lambda^2 +0.004 $$

los valores propios son: $$ \lambda = \pm\sqrt{0.004}i $$

son imaginarios puros, es decir Re$(\lambda) = 0$, asi que el pto crítico $(400, 5000)$ es un centro, el cual es simplemente estable.

## Modelo de competencia

$$ x^{\prime} = 0.004x(50-x-0.75y) $$

$$ y^{\prime} = 0.001y(100-y-3x) $$


Los ptos críticos son:

$$ (0,0), (0,100), (50,0), (20,40) $$

en este caso

$$ f(x,y) = 0.2x -0.004x^2 -0.003xy $$

y

$$ g(x,y) = 0.1y -0.001y^2 -0.003xy $$

las derivadas parciales son

$$ f_x(x,y)= 0.2-0.008x -0.003y $$

$$ f_y(x,y) = -0.003x $$

$$ g_x(x,y) = -0.003y $$

$$ g_y(x,y) = 0.1 - 0.002y -0.003x $$

La matriz Jacobiana es

$$ J(x,y) = \begin{bmatrix} 0.2-0.008x -0.003y & -0.003x \\ -0.003y  & 0.1 - 0.002y -0.003x\end{bmatrix} $$

Entonces la linealización alrededor del pto crítico $(0,0)$ es

$$ \begin{bmatrix} x \\y\end{bmatrix}^{\prime} = \begin{bmatrix} 0.2 & 0 \\ 0 & 0.1\end{bmatrix}\begin{bmatrix}x-0 \\y-0\end{bmatrix} $$

los valores propios son $\lambda =0.2 >0$ y $\lambda = 0.1 >0$, este es un repulsor, el cual es inestable.
