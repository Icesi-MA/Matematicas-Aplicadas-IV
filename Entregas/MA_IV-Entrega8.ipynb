3.   (20 ptos) Consider the nonlinear system of differential equations

\begin{align*}

\end{align*}

(b) Plot the phase portrait of this system in an applet. Do you
think that there is conserved quantity for this system? Can you find one?

\begin{align*}
\frac{dx}{dt} &= \frac{1-y^2}{1+y^2}\\
\frac{dy}{dt} &= x,\\
\end{align*}

Is this a Hamiltonian system?

(d) Can you now find a conserved quantity for the original system?
