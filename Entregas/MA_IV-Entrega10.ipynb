1.   (18 pts) Consider the system
\begin{align*}
\frac{dx}{dt} &= x\\
\frac{dy}{dt} &= -x-\frac{y}{4}+x^2,\\
\end{align*}

(a) Verify that the function
$$L(x, y) = \frac{y^2}{2}-\frac{x^2}{2}.-\frac{x^3}{3},$$
is a Lyapunov function for the system.

(b) Sketch the level sets of $L$.

(c) What can you conclude about the phase portrait of the system from the information in parts (a) and (b)? (Sketch the phase portrait and write a short essay describing what you know about the phase portrait and how you know it.)

2. (20 pts) Consider the damped pendulum system: \begin{align*}
\frac{d\theta}{dt} &= v\\
\frac{dv}{dt} &= -\frac{b}{m}v-\frac{g}{l}\sin\theta,\\
\end{align*}
where $b$ is the damping coefficient, $m$ is the mass of the bob, $l$ is the length of the arm, and $g$ is the acceleration of gravity ($g ≈ 9.8 m/s^2$). Suppose we take $l = 9.8$ m , $m = 1$, and $b$ large, say $b = 4$ or else. For the damped pendulum system above and with this choice of parameter values, do the following.

(a) Find the eigenvalues and eigenvectors of the linearized system at the equilibrium point $(0, 0)$.

(b) Find the eigenvalues and eigenvectors of the linearized system at the equilibrium point $(π, 0)$.

(c) Sketch the phase portrait near the equilibrium points.

(d) Sketch the entire phase portrait. [Hint: Begin by sketching the level sets of $H$ as in the text.]

3. (12 pts)  Consider the  Hamiltonian system:

\begin{align*}
\frac{dx}{dt} &= \frac{\partial H}{\partial y}\\
\frac{dy}{dt} &= -\frac{\partial H}{\partial x},\\
\end{align*}

and the gradient system

\begin{align*}
\frac{dy}{dt} &= \frac{\partial H}{\partial x}\\
\frac{dv}{dt} &= \frac{\partial H}{\partial y}.\\
\end{align*}

where $H$ is the same function for both systems. How are the two phase portraits related? Try to give a explicit example of a such a function $H$.
