**Problema 1**

Considere el sistema de ecuaciones diferenciales

$$ \dfrac{dx}{dt} = ax+by $$
$$ \dfrac{dy}{dt} = cx +ay $$

donde $a, b$ y $c$ son múmeros reales.



1.   Determine que condiciones deben satisfacer $a, b$ y $c$ de tal forma que el sistema tenga un solo equilibrio.
2.   ¿Qué condiciones deben satisfacer $a, b$ y $c$ para que $(0,0)$ sea una espiral estable?
3. ¿Qué condiciones deben satisfacer $a, b$ y $c$ para que $(0,0)$ sea un equilibrio asintoticamente estable?
4. ¿Qué condiciones deben satisfacer $a, b$ y $c$ para que $(0,0)$ sea un punto silla?

**Problema 2**

Un objeto que pesa 16 libras se une a un resorte de 5 pies de longitud. en la posición de equilibrio, el resorte mide 8.2 pies. Si el peso se eleva y se suelta del reposo en un punto a 2 pies arriba de la posición de equilibrio, determine el problema de valor inicial que modela la situación, conviertalo a un sistema de ecuaciones de primer orden y resuelvalo analíticamente y numéricamente.
