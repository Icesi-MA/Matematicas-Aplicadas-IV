## Problema

Considere la ecuación diferencial

$$\dfrac{d^2x}{dt^2} + ax - x^3 = 0$$

donde $a$ es un parámetro real.

1.   Escriba la ecuación diferencial como un sistema de ecuaciones de primer orden.
2.   Determine todos los valores de $a$ que hagan que el sistema tenga exactamente 3 puntos críticos.
3. Determine todos los puntos críticos del sistema si $a=9$, linealice el sistema en torno a cada punto crítico y haga un análisis de estabilidad para cada punto crítico.
4. Use el método de Euler trabajado en clase para resolver numéricamente el sistema para el valor de $a=9$ y compruebe lo encontrado en la pregunta anterior usando gráficas de las listas o variables encontradas al resolver numéricamente el sistema.
